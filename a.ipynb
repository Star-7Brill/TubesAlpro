```python
arr = [12, 11, 9, 26, 5]
def ins(arr):
    for i in range(1, len(arr)): #kenapa 1 karena mulai nya dari after 0
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j = j - 1
        arr[j + 1] = key
    return arr

sorted = ins(arr)
print(sorted)
```

```python
def sel(arr):
    n = len(arr)
    for i in range(n):
        minIdx = i
        for j in range(i + 1, n):
            if arr[j] < arr[minIdx]:
                minIdx = j 
        arr[i], arr[minIdx] = arr[minIdx], arr[i]
    return arr

arr = [5, 8, 3, 9, 1, 2]
sorted = sel(arr)
print(sorted)
```
