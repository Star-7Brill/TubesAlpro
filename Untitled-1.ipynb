```python
#Soal 1. Pemetaan Area Rawan Banjir
class AreaBanjir:

    def banjir(self):
        n, m = input().split()
        n = int(n)
        m = int(m)

        maks = -1
        mins = 100000

        maksBaris = 0
        maksKolom = 0

        minBaris = 0
        minKolom = 0

        for i in range(n):
            baris = input().split()
            for j in range(m):
                nilai = int(baris[j])
                if nilai > maks:
                    maks = nilai
                    maksBaris = i
                    maksKolom = j
                if nilai < mins:
                    mins = nilai
                    minBaris = i
                    minKolom = j
        print(f"Maksimun: {maks} di ({maksBaris}, {maksKolom})")
        print(f"Minimum: {mins} di ({minBaris}, {minKolom})")

data = AreaBanjir()
data.banjir()
```

Maksimun: 250 di (0, 3)
Minimum: 5 di (1, 1)

```python
class AreaBanjir:

    def proses(self):

        n, m = input().split()

        n = int(n)
        m = int(m)

        maks = -1
        minim = 100000

        maksBaris = 0
        maksKolom = 0

        minBaris = 0
        minKolom = 0

        for i in range(n):

            baris = input().split()

            for j in range(m):

                nilai = int(baris[j])

                if nilai > maks:
                    maks = nilai
                    maksBaris = i
                    maksKolom = j

                if nilai < minim:
                    minim = nilai
                    minBaris = i
                    minKolom = j

        print("Maksimum:", maks, "di", "(", maksBaris, ",", maksKolom, ")")
        print("Minimum:", minim, "di", "(", minBaris, ",", minKolom, ")")


data = AreaBanjir()
data.proses()
```

```python
#Soal 2. Analisis Nilai Ujian Siswa Time Limit
class nilaiSiswa:
    def ujian(self):
        n = int(input())
        data = input().split()

        maks = int(data[0])
        mins = int(data[0])

        total = 0

        for i in range(n):
            nilai = int(data[i])
            total = total + nilai

            if nilai > maks:
                maks = nilai
            if nilai < mins:
                mins = nilai
        rata = total / n
        print(f"Nilai maksimum: {maks}")
        print(f"Nilai minimum: {mins}")
        print(f"Rata-rata: {rata:.2f}")

data = nilaiSiswa()
data.ujian()
```

Nilai maksimum: 92
Nilai minimum: 78
Rata-rata: 86.60

```python
#Soal Tugas Pendahuluan Week 11
class SensorSuhu:
    def __init__(self, dataSuhu):
        self.dataSuhu = dataSuhu
    
    def maksimum(self):
        return max(self.dataSuhu)
    
    def minimum(self):
        return min(self.dataSuhu)
    
    def selisih(self):
        return self.maksimum() - self.minimum()
    
    def hasil(self):
        maks = self.maksimum()
        mins = self.minimum()
        selisih = self.selisih()
        print(f"{maks:.2f} {mins:.2f} {selisih:.2f}")

n = int(input())
inputData = input().split()
dataSuhu = []
for i in inputData:
    dataSuhu.append(float(i))

sensor = SensorSuhu(dataSuhu)
sensor.hasil()
```
